```python

import pandas as pd
import numpy as np
from scipy.stats import zscore

# Dataset containing an unusually high value
data = {
    "Hotel_Cost": [2000, 2200, 2100, 2300, 2050,
                   2150, 2250, 2200, 2100, 15000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# IQR method
Q1 = df["Hotel_Cost"].quantile(0.25)
Q3 = df["Hotel_Cost"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

iqr_outliers = df[
    (df["Hotel_Cost"] < lower_limit) |
    (df["Hotel_Cost"] > upper_limit)
]

print("\nOutliers using IQR:")
print(iqr_outliers)

# Z-score method
df["Z_Score"] = zscore(df["Hotel_Cost"])
z_outliers = df[df["Z_Score"].abs() > 3]

print("\nOutliers using Z-score:")
print(z_outliers)

# Remove IQR-detected outliers
cleaned_df = df[
    (df["Hotel_Cost"] >= lower_limit) &
    (df["Hotel_Cost"] <= upper_limit)
]

print("\nDataset after removing IQR outliers:")
print(cleaned_df)
```
