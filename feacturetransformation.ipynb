```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler



data = {
    "Destination": ["Goa", "Pune", "Manali", "Jaipur", "Mumbai"],
    "Hotel_Cost": [3000, 2000, 12000, 4000, 2500],
    "Travel_Cost": [1500, 800, 5000, 2000, 1000],
    "Days": [3, 2, 5, 4, 2]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)




scaler = MinMaxScaler()

features = df[["Hotel_Cost", "Travel_Cost", "Days"]]

scaled_data = scaler.fit_transform(features)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=["Hotel_Cost", "Travel_Cost", "Days"]
)

print("\nDataset After Min-Max Scaling:")
print(scaled_df)




standard_scaler = StandardScaler()

standard_data = standard_scaler.fit_transform(features)

standard_df = pd.DataFrame(
    standard_data,
    columns=["Hotel_Cost", "Travel_Cost", "Days"]
)

print("\nDataset After Standardization:")
print(standard_df)




df["Log_Hotel_Cost"] = np.log1p(df["Hotel_Cost"])

print("\nDataset After Log Transformation:")
print(df[["Hotel_Cost", "Log_Hotel_Cost"]])




df["Total_Cost"] = df["Hotel_Cost"] + df["Travel_Cost"]

df["Cost_Per_Day"] = df["Total_Cost"] / df["Days"]




print("\nFinal Dataset After Feature Engineering:")
print(df)



print("Feature transformation techniques applied successfully.")
print("New features Total_Cost and Cost_Per_Day created successfully.")
```
